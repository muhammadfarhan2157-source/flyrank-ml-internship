## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision: Which pages should a content reviewer look at first out of a large inventory, given limited review capacity?
Unit of analysis: one content page (content_id), scored per client (client-holdout, since pages from the same client shouldn't leak between train/test).
Who acts, and how: A content editor/reviewer works down a ranked queue and applies an action per page — refresh, expand, protect, prune, or monitor — starting from the top.
Cost of a wrong call: A false positive wastes reviewer hours on a page that didn't need attention. A false negative means a real problem page (declining, thin, stale) sits unreviewed while it keeps losing visibility — a missed opportunity cost, not caught until later.
Why ML, not a plain rule: The starter pipeline already shows a hand-written rule (a weighted sum of visibility, freshness, position, and depth signals) captures real pattern but leaves a lot on the table — Precision@50 of 0.24 vs. 0.74 for a learned model on the same inputs. That gap suggests the true pattern is tangled across many signals in a way a simple if-statement can't fully capture.
