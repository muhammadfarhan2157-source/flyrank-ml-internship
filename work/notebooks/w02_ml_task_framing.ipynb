## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")
df = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)].drop_duplicates(subset="content_id")

df["is_declining_label"] = (df["trend_direction"] == "down").astype(int)
print("Rows:", len(df))
print("Share declining:", round(df["is_declining_label"].mean(), 3))
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
def precision_at_k(y_true, y_score, k=50):
    top_k = pd.Series(y_score).nlargest(k).index
    return y_true.iloc[top_k].mean()

print("Metric: precision@50 — fraction of top-50 ranked pages that match the label")
```
